# **Calcul des moyennes des UE**

Version 1 : le calcul des moyennes des UE est réalisé avec des notes et coefficients saisis au clavier.

```python
cache=float(0) #I define all variables
note=float(0)
coeff=float(0)
compteur=int(101) #set compteur to 101 for R 101-115
resultat=float(0)

while compteur != 116: #first while to compute the coeff of the R's
    addition=float(input("saisisez votre note pour R{}".format(compteur))) #ask grade
    coeff=float(input("saisisez votre coeff pour R{}".format(compteur))) #ask coeff
    compteur=compteur+1 #increase by 1 R
    cache=cache+(addition*coeff) #put in cache the result
    
compteur=11 #set compteur to 11 for the SAE 11-16
while compteur != 17: #second while to compute the coeff of the SAE's
    addition=float(input("saisisez votre note pour SAE{}".format(compteur))) #ask grade
    coeff=float(input("saisisez votre coeff pour SAE{}".format(compteur))) #ask coeff
    compteur=compteur+1 #increase by 1 R
    cache=cache+(addition*coeff) #add to cache the result
resultat=cache/21 #divide the cache by the number of grades
resultat #show the final result
```

Version 2 : le calcul des moyennes des UE est réalisé avec des notes et des coefficients dans des listes.

```python
import numpy as np

matiere=[f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range (11,17)]#crée deux liste, liste R101-115 et SAE11-16

note=[float(input(f"quelle est la note pour {elt} ?")) for elt in matiere]#demande une note pour chaque matiere et l'associe à celle ci

coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]#liste des coeffs ue1
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]#ue2
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]#ue3

moy1=round(np.average(note,weights=coef1),2)#calcul moy ue1
moy2=round(np.average(note,weights=coef2),2)#ue2
moy3=round(np.average(note,weights=coef3),2)#ue3
print("Moyenne de l'UE1",moy1, "Moyenne de l'UE2",moy2, "Moyenne de l'UE3",moy3)# affiche 
```

Version 3 : le calcul des moyennes des UE est réalisé avec des coefficients et des notes dans des dictionnaires.

```python
matiere=[f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range (11,17)]#crée deux liste, liste R101-115 et SAE11-16

coef1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]#liste des coeffs ue1
coef2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]#ue2
coef3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]#ue3

notes=[float(input(f"quelle est la note pour {elt} ?")) for elt in matiere]

dicnote = {cle : valeur for cle,valeur in zip(matiere,notes)}
dicoefue1 = {cle : valeur for cle,valeur in zip(matiere,coef1)}
dicoefue2 = {cle : valeur for cle,valeur in zip(matiere,coef2)}
dicoefue3 = {cle : valeur for cle,valeur in zip(matiere,coef3)}

sommeN=0
sommeC=0
for elt in matiere:
    sommeN += dicnote[elt]*dicoefue1[elt]
    sommeC += dicoefue1[elt]
moy1=round(sommeN/sommeC,2)
sommeN=0
sommeC=0
for elt in matiere:
    sommeN += dicnote[elt] * dicoefue2[elt]
    sommeC += dicoefue2[elt]
moy2=round(sommeN/sommeC,2)
sommeN=0
sommeC=0
for elt in matiere:
    sommeN += dicnote[elt] * dicoefue3[elt]
    sommeC += dicoefue3[elt]
moy3=round(sommeN/sommeC,2)
print(f"la moyenne de l'ue1 est {moy1}")
print(f"la moyenne de l'ue1 est {moy2}")
print(f"la moyenne de l'ue1 est {moy3}")
```

Version 4 : on ajoute une partie de visualisation graphique des résultats et un commentaire sur la décision jury.

```python
import matplotlib.pyplot as plt

# 1. Création des listes de matières et des coefficients
matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

coef1 = [10, 10, 7, 7, 0, 5, 0, 6, 0, 5, 4, 2, 5, 5, 0, 20, 20, 0, 0, 0, 7]
coef2 = [4, 0, 2, 8, 6, 0, 0, 0, 0, 5, 5, 2, 9, 9, 3, 0, 0, 29, 0, 0, 7]
coef3 = [4, 0, 2, 0, 0, 5, 15, 6, 4, 5, 5, 2, 0, 0, 3, 0, 0, 0, 20, 20, 7]

# 2. Saisie des notes par l'utilisateur
notes = [float(input(f"Quelle est la note pour {elt} ? ")) for elt in matiere]

# 3. Création des dictionnaires
dicnote = {cle: valeur for cle, valeur in zip(matiere, notes)}
dicoefue1 = {cle: valeur for cle, valeur in zip(matiere, coef1)}
dicoefue2 = {cle: valeur for cle, valeur in zip(matiere, coef2)}
dicoefue3 = {cle: valeur for cle, valeur in zip(matiere, coef3)}

# 4. Calcul de la moyenne pour l'UE 1
sommeN = 0
sommeC = 0
for elt in matiere:
    sommeN += dicnote[elt] * dicoefue1[elt]
    sommeC += dicoefue1[elt]
moy1 = round(sommeN / sommeC, 2) if sommeC > 0 else 0

# 5. Calcul de la moyenne pour l'UE 2
sommeN = 0
sommeC = 0
for elt in matiere:
    sommeN += dicnote[elt] * dicoefue2[elt]
    sommeC += dicoefue2[elt]
moy2 = round(sommeN / sommeC, 2) if sommeC > 0 else 0

# 6. Calcul de la moyenne pour l'UE 3
sommeN = 0
sommeC = 0
for elt in matiere:
    sommeN += dicnote[elt] * dicoefue3[elt]
    sommeC += dicoefue3[elt]
moy3 = round(sommeN / sommeC, 2) if sommeC > 0 else 0

# 7. Attribution automatique des couleurs et des messages du jury
ues = ['UE 1', 'UE 2', 'UE 3']
mes_moyennes = [moy1, moy2, moy3]
couleurs = []
messages_jury = []

for moy in mes_moyennes:
    if moy >= 10:
        couleurs.append('g')  # Vert
        messages_jury.append("Validé (Bravo !)")
    elif 8 <= moy <= 10:
        couleurs.append('orange')  # Orange
        messages_jury.append("Passable / Rattrapage possible")
    else:
        couleurs.append('r')  # Rouge
        messages_jury.append("Non validé (Ajourné)")

# Affichage des résultats et des messages du jury dans la console
print("\RÉSULTATS ET AVIS DU JURY")
for ue, moy, msg in zip(ues, mes_moyennes, messages_jury):
    print(f"La moyenne de l'{ue} est {moy}/20 -> Avis : {msg}")

# 8. Génération du graphique à barres personnalisé
plt.figure(figsize=(8, 5))
bars = plt.bar(ues, mes_moyennes, color=couleurs, width=0.5)

plt.xlabel('Unités d\'Enseignement', fontsize=12)
plt.ylabel('Moyenne (sur 20)', fontsize=12)
plt.title('Mes moyennes par Unité d\'Enseignement', fontsize=14, fontweight='bold')
plt.ylim(0, 20)

plt.tight_layout()
plt.show()
```
